## 1. Load the Data

The first step is to read the CSV file into pandas so the data can be explored and analyzed. This allows the airline arrival information to be structured into a DataFrame for easier comparison.

```python
import pandas as pd

df = pd.read_csv("Project01.csv")

df
```

## Clean up Repeated Airline Labels

Some airline names are blank in rows where the label should carry down.  
I’m using forward fill so every row keeps the right airline name.

**Key point:** this makes grouping by airline possible later. (easier)

```python
df["Airline"] = df["Airline"].ffill()

df
```

## Focus on Delayed Flights Only

Now I’m filtering to just rows where `Status == "delayed"`.

**Key point:** I've isolated the records we need for delay comparisons.

```python
delays = df[df["Status"] == "delayed"]

delays
```

## 4. Calculating Total Delays

The next step is to calculate the `Total Delays` of flights across all five destinations for each airline. Summing these values provides a direct comparison of overall arrival delays.

**Key point:** this creates a single number we can aggregate by airline.

```python
delays["Total Delays"] = delays.iloc[:, 2:].sum(axis=1)

delays[["Airline", "Total Delays"]]
```

## 5. Comparing Delay Rates

The raw delay totals help, but rates are fairer since airlines may have different flight volumes.  
Here I calculate total flights, total delays, then delay rate percentage.

**Key point:** delay rate gives the more balanced comparison.


```python
df["Total Flights"] = df.iloc[:, 2:].sum(axis=1)

totalFlights = df.groupby("Airline")["Total Flights"].sum()
totalDelays = delays.groupby("Airline")["Total Delays"].sum()

delayRate = (totalDelays / totalFlights) * 100

delayRate
```

## Final Thoughts

After cleaning the provided data, I focused on delayed arrivals to compare performance between the two airlines. AM WEST had a higher total number of delays across the five destinations with `787 total delays`. However, looking at delay rates instead of just raw totals gave a clearer picture of reliability. When the number of delayed flights is considered in relation to total flights, the comparison feels more grounded. Overall, the analysis suggests that proportional performance tells a more meaningful story than simple counts alone.
